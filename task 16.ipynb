```python
import pandas as pd


file_path = r'C:\Users\SK\Downloads\Iris'


try:
    df = pd.read_csv(file_path)
except Exception:
    df = pd.read_csv(file_path + '.csv')

print("Dataset successfully loaded!")
print(f"Shape of Dataset: {df.shape}")
df.head()
```

```python
print("=== 5 loc[] Examples (Label-based) ===")

# 1. Single row by label
print("\n1. Row at label 0:")
display(df.loc[0])

# 2. Slice of rows and specific columns by name
print("\n2. Rows 0 to 4 with specific columns:")
display(df.loc[0:4, ['SepalLengthCm', 'SepalWidthCm', 'Species']])

# 3. Column slice from SepalLengthCm to PetalWidthCm
print("\n3. Column slice (SepalLengthCm to PetalWidthCm):")
display(df.loc[0:2, 'SepalLengthCm':'PetalWidthCm'])

# 4. Conditional filtering with loc
print("\n4. Filter rows where Species is 'Iris-setosa':")
display(df.loc[df['Species'] == 'Iris-setosa', ['Id', 'Species', 'PetalLengthCm']].head(3))

# 5. Specific non-consecutive rows and columns
print("\n5. Specific rows [2, 5, 10] and columns:")
display(df.loc[[2, 5, 10], ['Id', 'SepalLengthCm']])
```

```python
print("=== 5 iloc[] Examples (Position-based) ===")

# 1. Single row by integer index
print("\n1. Row at position 0:")
display(df.iloc[0])

# 2. Positional slicing for rows and columns
print("\n2. Rows 0-4 and columns 0-2:")
display(df.iloc[0:5, 0:3])

# 3. Specific row indices and column indices
print("\n3. Specific rows [1, 3, 5] and columns [1, 4]:")
display(df.iloc[[1, 3, 5], [1, 4]])

# 4. Negative indexing for the last column
print("\n4. Last column (Species) for first 5 rows:")
display(df.iloc[0:5, -1])

# 5. Step size slicing (every 30th row)
print("\n5. Every 30th row using step size:")
display(df.iloc[0:150:30, 0:4])
```

```python
comparison_data = {
    "Feature": ["Selection Basis", "Boundary Inclusion", "Column Reference", "Handling Data Changes"],
    "loc[]": ["Uses labels (index names or column names)", "Both start and end labels are inclusive", "Requires exact column name strings", "Safe against column reordering/insertion"],
    "iloc[]": ["Uses integer positions (0-indexed)", "End index is exclusive (like Python slicing)", "Requires integer column indices", "Breaks if column positions change"]
}

df_comparison = pd.DataFrame(comparison_data)
display(df_comparison)
```
